```python
# Make a GET
req = requests.get("http://www.ilga.gov/senate/default.asp")
# leer el contenido del servidor 
src = req.text
# ver la salida
print(src[:1000])
```

```python
# Parsear
soup = BeautifulSoup(src, "lxml")
# Vista
print(soup.prettify()[:1000])
```

```python
# busca los tag que inician con “a”
a_tags = soup.find_all("a")
print(a_tags[:10])
```

```python
# tags "a" en el menu lateral
side_menus = soup("a", class_="sidemenu")
side_menus[:5]
```
